### Задание 1. Средний возраст пассажиров и стандартное отклонение (bottleneck)
Функции bn.nanmean() и bn.nanstd() умеют работать с пропусками (NaN) в столбце Age.

```python
# подключаем нужные библиотеки
import pandas as pd
import numpy as np
import bottleneck as bn

# снова читаем датасет, но уже через pandas
df = pd.read_csv('train.csv')

# берем столбец Age и превращаем его в обычный массив
age = df['Age'].to_numpy()

# считаем среднее и стандартное отклонение
print('Средний возраст пассажиров:')
print(bn.nanmean(age))

print('Стандартное отклонение возраста:')
print(bn.nanstd(age))
```

## Часть 3. Оптимизация типов данных в pandas

### Задание 1. Читаем датасет Housing.csv с помощью pandas

```python
# читаем файл о ценах на дома
housing = pd.read_csv('Housing.csv')

# смотрим на первые строки
housing.head()
```

### Задание 2. Какие типы данных оптимальны для каждого столбца (выводы)
Посмотрим, какие бывают типы данных в датасете и какие значения хранятся в столбцах.

```python
# смотрим типы данных всех столбцов
print(housing.dtypes)
```

Мои выводы, какой тип лучше всего подходит каждому столбцу:

- **price** - это целые числа от 1 750 000 до 13 300 000. Максимум меньше 2 147 483 647, поэтому хватит самого маленького целого типа `int32`, который занимает 4 байта вместо 8.
- **area** - целые числа от 1 650 до 16 200. Максимум меньше 32 767, поэтому подойдет `int16` (2 байта).
- **bedrooms** (1-6), **bathrooms** (1-4), **stories** (1-4), **parking** (0-3) - значения совсем маленькие, поэтому возьмем тип `int8`, который занимает 1 байт.
- **mainroad, guestroom, basement, hotwaterheating, airconditioning, prefarea** - в этих столбцах только два значения `yes` и `no`. Это категориальные данные, для них лучше тип `category`.
- **furnishingstatus** - содержит всего три значения (`furnished`, `semi-furnished`, `unfurnished`). Это тоже категория, поэтому тип `category`.

В итоге: числовые столбцы с большими числами уменьшаем до int32/int16/int8, а строковые категориальные столбцы переводим в type `category`.

### Задание 3. Меняем типы данных и сравниваем память до и после

```python
# смотрим, сколько памяти занимает датасет сейчас
memory_before = housing.memory_usage(deep=True).sum() / 1024
print('Память до оптимизации:', round(memory_before, 2), 'КБ')

# меняем типы данных у числовых столбцов
housing['price'] = housing['price'].astype('int32')
housing['area'] = housing['area'].astype('int16')
housing['bedrooms'] = housing['bedrooms'].astype('int8')
housing['bathrooms'] = housing['bathrooms'].astype('int8')
housing['stories'] = housing['stories'].astype('int8')
housing['parking'] = housing['parking'].astype('int8')

# меняем типы данных у категориальных столбцов
housing['mainroad'] = housing['mainroad'].astype('category')
housing['guestroom'] = housing['guestroom'].astype('category')
housing['basement'] = housing['basement'].astype('category')
housing['hotwaterheating'] = housing['hotwaterheating'].astype('category')
housing['airconditioning'] = housing['airconditioning'].astype('category')
housing['prefarea'] = housing['prefarea'].astype('category')
housing['furnishingstatus'] = housing['furnishingstatus'].astype('category')

# смотрим, сколько памяти занимает датасет после оптимизации
memory_after = housing.memory_usage(deep=True).sum() / 1024
print('Память после оптимизации:', round(memory_after, 2), 'КБ')

# считаем, сколько памяти удалось сэкономить
print('Сэкономлено:', round(memory_before - memory_after, 2), 'КБ')

print()
print('Проверяем новые типы данных:')
print(housing.dtypes)
```
